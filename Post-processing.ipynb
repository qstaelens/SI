```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# Data
# ============================================================

bond_dim = np.array([37,72,137,274])
Ms = np.array([0.076695,0.052756,0.036697,0.02472])

#Interchain interactions 
bond_dim1 = np.array([37,62,107,176])
Ms1 = np.array([0.090405, 0.07761, 0.0717145,0.069869])

# ============================================================
# Convert to 1/D
# ============================================================

invD = 1.0 / bond_dim
invD1 = 1.0 / bond_dim1

# ============================================================
# Linear extrapolation using largest D points
# ============================================================

nfit = 9  # number of largest-D points used for fit

# No interchain interactions
coef, cov = np.polyfit(invD[-nfit:], Ms[-nfit:], 1, cov=True)
fit_fn = np.poly1d(coef)
Ms_inf = fit_fn(0)
Ms_inf_err = 3*np.sqrt(cov[1,1])

# Interchain interactions
coef1, cov1 = np.polyfit(invD1[-nfit:], Ms1[-nfit:], 1, cov=True)
fit_fn1 = np.poly1d(coef1)
Ms1_inf = fit_fn1(0)
Ms1_inf_err = 3*np.sqrt(cov1[1,1])

print(f"Extrapolated Ms(D→∞) without interchain interactions = {Ms_inf:.6f} ± {Ms_inf_err:.6f}")
print(f"Extrapolated Ms(D→∞) with interchain interactions    = {Ms1_inf:.6f} ± {Ms1_inf_err:.6f}")

# ============================================================
# Smooth fit lines
# ============================================================

xfit = np.linspace(0, max(invD1[-nfit:])*1.05, 300)

# ============================================================
# Plot
# ============================================================

plt.figure(figsize=(6,4))

# Data
line0, = plt.plot(invD[-nfit:], Ms[-nfit:], 'o', label='No interchain interactions')
line1, = plt.plot(invD1[-nfit:], Ms1[-nfit:], 's', label='Interchain interactions')

# Fits
plt.plot(xfit, fit_fn(xfit), linestyle='--', linewidth=2, color=line0.get_color())
plt.plot(xfit, fit_fn1(xfit), linestyle='--', linewidth=2, color=line1.get_color())

# Labels
plt.xlabel(r"$1/D$", fontsize=14)
plt.ylabel(r"$M_s$", fontsize=14)

#plt.xscale('log')
plt.tick_params(axis='both', which='major', labelsize=12)

plt.grid(True)
plt.legend(fontsize=11)

plt.xlim(left=0)
plt.ylim(bottom=0, top = 0.1)

plt.tight_layout()
plt.show()
```
